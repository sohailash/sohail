```python
"""
NowCerts API - Flexible Customer Search Script
Saves files with timestamps to prevent overwriting
Author: Ranveer Verma
Date: October 2025
"""

import requests
import json
import csv
from typing import Dict, List, Optional
from datetime import datetime
import os

# =============================================================================
# CONFIGURATION - YOUR API CREDENTIALS
# =============================================================================

ACCESS_TOKEN = "W-WZD-Aw3_JaJzjR7Xicx9zMrXtpYj16XnjHFiu_WfqLwJE6MulEyg-3c7UoGtF_QxvSXn0hBpYmSRio6LdyDgqLw2ocWJYrlzEYzZfWkySPqfFH_huX9QSX1xjweIQABYcqDO9BuDR4HlQauHBw1EClan9fwiitlhwlTOHJEZrTFlNLuqudPFUsa2xUqod9Ues-Lyl5AGwU8NDLjomQZhUIKmHalRXI-y7upIudNauCNx9VrcMuA3e82Zke2047Xkdeo9-mFZhFiOzNmzao9GdGKx9YmqxmNiabjSDofsxrIIJ6hrQB4O63D63PblJrQfg-OD5G06mAf2XhUGhrOP5YT3SYsLMhZMmfqpMj_5xyrYIVCfb2oziZacVAbtIs1OXCZ_DstedZNqR2eli8s_h_9B-QJuqkxAdJSgWUtsVQHBPrmLDnGAYiYPrduE7OHzQuFB7_0VZqnSmjEAFEEpzGvtMkn5wGipfBXc4IEzrJbVX19iiz7Eu1TT7yA2jtEeV8d8xy24IZ94vRBj-KyRqeafmtrej9pTcmuAQlHAibxDClfzjuHACFwj8JuS27odH23pz5rpYknYdHbLoareO-0rvGjSOPHB9upvqyjQKIblxYwIIhslOTsix3_k0-7eSF0gmbCcs07r4QZ2_BWzERXxW7fInGc6zZitPzGEYKi1gzjWKkWLdwjPipWxzOKZmGx3IDwBv2M9fwAO4nHMwqFzca1Nm7I9dh0t-xOpwAxbTS7_o-5sbBLKuV8a0Bk1YzMPaW31kRJIGF0YrTaswF9tTfNbTKVQKYXCvXydEkLf5CZ6taiogdJsgrX5-EwguRu_NuJZ7ueTzkhb68ZJkv-x-w0cUaj-Pl80VUob4a55FkclIO7tMJSYWPEh4DFpzwN4Y_5pj4ttGByRNji8SbMUi4maA4juBHgEWKxy5uansaMcLlMB_YlQOB9j_9kDK8Curp2O4w_dqlUtSC5vgVqimufWlRbbnXdu-jvTQQlFXzO7aYZbbQMh5DCbPf5sgm_MdnQ-KA4KBCRZQ1zZ_HzJ9KYf1T7w3vcpeK4_AH6K7xygZlbV6d1Kxzn04On3cvP-Fm8t-3aA1B-doLYnGq8GYd0LOV21ouwBLyLbw02hmdISaZpskUvpIxK-RVPATmGjMSqH9OxRLGeXo40VRCE5KRjbPUDAPmLBsh9DgXtOBwbSOv7pfgJwR9DRqM4jxVSlIEhrykqs6qrfXCj5PYI9Kbbgk29oqeBoggGL7Gm-S9IzDa_wuaRP8JoQ11LSQvBQrOkycVND2AZygYDRySPheP5pQ5IpRxKnWu7GDobnep_3njferMPQHM2hIyM7gexJh6jfQbvkSE_ngFtsEwgD2RPBPOBShRsEqfmOZ9I1Y02PaCTZAnQ-UXlEKYSmYPOdUu24drn-ixldGbTlvGEPRsABBUvWG1rzydRUYKOvsyZuLWkwK31UQaDCZ7uO0WwCFO6G2yjJSAyMtdCBFU2t5df9U0eOdnrD00fbEQlrbiodb4WMUU84W2GIfzl9uUXRTApSU8L9qnkqUWxPCSDweXnYDQGGZ2fvsa3AsQjP3HCSLZXXxZ5uHLRhB8JXiIQEa5kSUZUPJo2SxYEfzF-qHPK4e016h5tUbHlOcA8w4ef_VPztH2Kc_3n5IgV7jFlzPppU9vbRwaJAYRV51yqn9aLBKnRGC4Va6964HODFoY0M-4084m0pM-YWh0pnbebeRoUXKF1lRJRzDJDQ"
REFRESH_TOKEN = "EC0C0FD0AD727600112248118DECE0E36CF5A6B4FE2FCFDBCAD93AEB4A18D348"

BASE_URL = "https://api.nowcerts.com/api"
REFRESH_URL = "https://identity.nowcerts.com/connect/token"
CLIENT_ID = "nowcerts_public_api"

# =============================================================================
# SEARCH CRITERIA - CHANGE THESE TO FILTER YOUR RESULTS
# =============================================================================

SEARCH_CRITERIA = {
    "Name": "A",
    "Address": "",
    "Email": "",
    "Phone": "",
    "InsuredId": "",
    "CustomerId": "",
    "DatabaseId": "",
}

# =============================================================================
# FILTERING OPTIONS - APPLIED AFTER DATA IS RETRIEVED
# =============================================================================

FILTER_OPTIONS = {
    "active_only": True,
    "city": "",
    "state": "",
    "customer_type": None,
}

# =============================================================================
# OUTPUT OPTIONS
# =============================================================================

OUTPUT_OPTIONS = {
    # Save as JSON? (True/False)
    "save_json": True,
    
    # Save as CSV? (True/False)
    "save_csv": True,
    
    # Use timestamped filenames? (True/False)
    # True = nowcerts_customers_20251009_013000.json (never overwrites)
    # False = nowcerts_customers.json (overwrites each time)
    "use_timestamp": True,
    
    # Base filename (timestamp will be added if use_timestamp=True)
    "base_filename": "nowcerts_customers",
    
    # Print summary to console? (True/False)
    "print_summary": True,
    
    # Maximum number of records to display in console (0 = show all)
    "max_display": 10,
}

# =============================================================================
# TOKEN MANAGEMENT
# =============================================================================

def refresh_access_token():
    """Refresh the access token using refresh token"""
    print("\n🔄 Refreshing access token...")
    
    data = {
        "grant_type": "refresh_token",
        "refresh_token": REFRESH_TOKEN,
        "client_id": CLIENT_ID,
    }
    
    try:
        response = requests.post(REFRESH_URL, data=data, timeout=30)
        
        if response.status_code == 200:
            tokens = response.json()
            new_access_token = tokens.get("access_token")
            new_refresh_token = tokens.get("refresh_token")
            
            print("✅ Token refreshed successfully")
            return new_access_token, new_refresh_token
        else:
            print(f"❌ Token refresh failed: HTTP {response.status_code}")
            print(f"   Response: {response.text[:200]}")
            return None, None
            
    except Exception as e:
        print(f"❌ Error during token refresh: {str(e)}")
        return None, None

# =============================================================================
# API DATA RETRIEVAL
# =============================================================================

def build_api_url():
    """Build API URL with search parameters"""
    base = f"{BASE_URL}/Customers/GetCustomers"
    
    params = []
    for key, value in SEARCH_CRITERIA.items():
        if value:
            params.append(f"{key}={value}")
    
    if params:
        return f"{base}?{'&'.join(params)}"
    return base

def get_customers(access_token):
    """Retrieve customers from NowCerts API"""
    endpoint = build_api_url()
    
    print(f"\n📡 API Request:")
    print(f"   Endpoint: {endpoint}")
    
    headers = {
        "Authorization": f"Bearer {access_token}",
        "Content-Type": "application/json",
        "Accept": "application/json"
    }
    
    try:
        response = requests.get(endpoint, headers=headers, timeout=30)
        
        if response.status_code == 401:
            print("\n⚠️  Token expired, refreshing...")
            
            new_token, new_refresh = refresh_access_token()
            
            if new_token:
                global ACCESS_TOKEN, REFRESH_TOKEN
                ACCESS_TOKEN = new_token
                if new_refresh:
                    REFRESH_TOKEN = new_refresh
                
                print("🔁 Retrying request...")
                headers["Authorization"] = f"Bearer {new_token}"
                response = requests.get(endpoint, headers=headers, timeout=30)
            else:
                print("❌ Cannot refresh token")
                return []
        
        if response.status_code == 200:
            data = response.json()
            
            if isinstance(data, list):
                print(f"✅ Retrieved {len(data)} customer(s) from API")
                return data
            else:
                print(f"⚠️  Unexpected response format: {type(data)}")
                return []
        else:
            print(f"❌ API Error: HTTP {response.status_code}")
            print(f"   Response: {response.text[:200]}")
            
            if response.status_code == 400:
                print("\n💡 TIP: Add at least one search parameter in SEARCH_CRITERIA")
            
            return []
            
    except Exception as e:
        print(f"❌ Error during API call: {str(e)}")
        return []

# =============================================================================
# DATA FILTERING
# =============================================================================

def apply_filters(customers):
    """Apply post-retrieval filters to customer data"""
    if not customers:
        return customers
    
    filtered = customers
    original_count = len(customers)
    
    print("\n🔍 Applying filters...")
    
    # Filter by active status
    if FILTER_OPTIONS["active_only"] is not None:
        filtered = [c for c in filtered if c.get("active") == FILTER_OPTIONS["active_only"]]
        status = "active" if FILTER_OPTIONS["active_only"] else "inactive"
        print(f"   ✓ Filtered by status ({status}): {len(filtered)} records")
    
    # Filter by city - handles None values
    if FILTER_OPTIONS["city"]:
        city_filter = FILTER_OPTIONS["city"].lower()
        filtered = [c for c in filtered 
                   if c.get("city") and c.get("city").lower() == city_filter]
        print(f"   ✓ Filtered by city ({FILTER_OPTIONS['city']}): {len(filtered)} records")
    
    # Filter by state - handles None values
    if FILTER_OPTIONS["state"]:
        state_filter = FILTER_OPTIONS["state"].lower()
        filtered = [c for c in filtered 
                   if c.get("stateNameOrAbbreviation") and 
                   c.get("stateNameOrAbbreviation").lower() == state_filter]
        print(f"   ✓ Filtered by state ({FILTER_OPTIONS['state']}): {len(filtered)} records")
    
    # Filter by customer type
    if FILTER_OPTIONS["customer_type"] is not None:
        filtered = [c for c in filtered if c.get("type") == FILTER_OPTIONS["customer_type"]]
        type_name = "Insured" if FILTER_OPTIONS["customer_type"] == 0 else "Prospect"
        print(f"   ✓ Filtered by type ({type_name}): {len(filtered)} records")
    
    if len(filtered) < original_count:
        print(f"\n📊 Filter Results: {original_count} → {len(filtered)} records")
    else:
        print(f"   No filters applied (using all {len(filtered)} records)")
    
    return filtered

# =============================================================================
# FILENAME GENERATION
# =============================================================================

def generate_filename(extension):
    """
    Generate filename with or without timestamp based on settings
    
    Args:
        extension: File extension ('json' or 'csv')
    
    Returns:
        Filename string
    """
    base = OUTPUT_OPTIONS["base_filename"]
    
    if OUTPUT_OPTIONS["use_timestamp"]:
        # Create timestamp: YYYYMMDD_HHMMSS
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        return f"{base}_{timestamp}.{extension}"
    else:
        # Simple filename (will overwrite)
        return f"{base}.{extension}"

# =============================================================================
# DATA EXPORT
# =============================================================================

def save_to_json(customers):
    """Save customers to JSON file with optional timestamp"""
    filename = generate_filename("json")
    
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(customers, f, indent=2, ensure_ascii=False)
        
        print(f"\n💾 JSON saved: {filename}")
        return filename
    except Exception as e:
        print(f"❌ Error saving JSON: {str(e)}")
        return None

def save_to_csv(customers):
    """Save customers to CSV file with optional timestamp"""
    if not customers:
        print("⚠️  No data to save to CSV")
        return None
    
    filename = generate_filename("csv")
    
    try:
        fieldnames = [
            'databaseId', 'insuredId', 'customerId',
            'commercialName', 'firstName', 'lastName',
            'email', 'phone', 'cellPhone', 'fax',
            'addressLine1', 'addressLine2', 'city', 
            'stateNameOrAbbreviation', 'zipCode',
            'active', 'type', 'website', 'fein', 'description'
        ]
        
        with open(filename, 'w', newline='', encoding='utf-8') as csvfile:
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames, extrasaction='ignore')
            writer.writeheader()
            
            for customer in customers:
                row = {**customer}
                if 'eMail' in row:
                    row['email'] = row['eMail']
                
                writer.writerow(row)
        
        print(f"💾 CSV saved: {filename}")
        return filename
        
    except Exception as e:
        print(f"❌ Error saving CSV: {str(e)}")
        return None

# =============================================================================
# DATA DISPLAY
# =============================================================================

def print_summary(customers):
    """Print formatted summary of customer data"""
    if not customers:
        print("\n📋 No customers to display")
        return
    
    max_display = OUTPUT_OPTIONS["max_display"]
    display_count = len(customers) if max_display == 0 else min(max_display, len(customers))
    
    print("\n" + "="*80)
    print("CUSTOMER DATA SUMMARY")
    print("="*80)
    
    for i, customer in enumerate(customers[:display_count], 1):
        print(f"\n{'─'*80}")
        print(f"📌 Customer #{i}")
        print(f"{'─'*80}")
        
        print(f"DatabaseId:   {customer.get('databaseId', 'N/A')}")
        print(f"InsuredId:    {customer.get('insuredId', 'N/A')}")
        print(f"CustomerId:   {customer.get('customerId', 'N/A')}")
        
        if customer.get('commercialName'):
            print(f"Business:     {customer.get('commercialName')}")
        else:
            first = customer.get('firstName', '')
            last = customer.get('lastName', '')
            if first or last:
                print(f"Name:         {first} {last}".strip())
        
        print(f"Email:        {customer.get('eMail', 'N/A')}")
        print(f"Phone:        {customer.get('phone', 'N/A')}")
        print(f"Cell:         {customer.get('cellPhone', 'N/A')}")
        
        addr1 = customer.get('addressLine1', '')
        if addr1:
            print(f"Address:      {addr1}")
        addr2 = customer.get('addressLine2', '')
        if addr2:
            print(f"              {addr2}")
        
        city = customer.get('city', '')
        state = customer.get('stateNameOrAbbreviation', '')
        zip_code = customer.get('zipCode', '')
        if city or state or zip_code:
            print(f"Location:     {city}, {state} {zip_code}".strip(", "))
        
        active_status = "✓ Active" if customer.get('active') else "✗ Inactive"
        cust_type = "Insured" if customer.get('type') == 0 else "Prospect"
        print(f"Status:       {active_status} | Type: {cust_type}")
    
    if len(customers) > display_count:
        remaining = len(customers) - display_count
        print(f"\n{'─'*80}")
        print(f"... and {remaining} more customer(s) (not displayed)")
    
    print("\n" + "="*80)
    print(f"Total Results: {len(customers)} customer(s)")
    print("="*80)

# =============================================================================
# MAIN EXECUTION
# =============================================================================

def main():
    """Main execution function"""
    
    print("="*80)
    print("NowCerts Customer Search - Flexible Criteria")
    print("="*80)
    print(f"\nTimestamp: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Show filename mode
    if OUTPUT_OPTIONS["use_timestamp"]:
        print("📁 File Mode: Timestamped (files will NOT be overwritten)")
    else:
        print("📁 File Mode: Simple (files will be overwritten)")
    
    print("\n📋 Active Search Criteria:")
    has_criteria = False
    for key, value in SEARCH_CRITERIA.items():
        if value:
            print(f"   • {key}: {value}")
            has_criteria = True
    
    if not has_criteria:
        print("   • No API search criteria")
        print("   • ⚠️  API requires at least one search parameter")
    
    print("\n🔍 Active Filters:")
    has_filters = False
    if FILTER_OPTIONS["active_only"] is not None:
        status = "Active" if FILTER_OPTIONS["active_only"] else "Inactive"
        print(f"   • Status: {status} only")
        has_filters = True
    if FILTER_OPTIONS["city"]:
        print(f"   • City: {FILTER_OPTIONS['city']}")
        has_filters = True
    if FILTER_OPTIONS["state"]:
        print(f"   • State: {FILTER_OPTIONS['state']}")
        has_filters = True
    if FILTER_OPTIONS["customer_type"] is not None:
        type_name = "Insured" if FILTER_OPTIONS["customer_type"] == 0 else "Prospect"
        print(f"   • Type: {type_name}")
        has_filters = True
    
    if not has_filters:
        print("   • No filters (showing all results)")
    
    print("\n" + "-"*80)
    print("STEP 1: Retrieving Data from API")
    print("-"*80)
    
    customers = get_customers(ACCESS_TOKEN)
    
    if not customers:
        print("\n❌ No data retrieved.")
        print("\n💡 SOLUTION: Add at least one search criterion")
        print("   Example: SEARCH_CRITERIA = {'Name': 'Smith', ...}")
        return
    
    print("\n" + "-"*80)
    print("STEP 2: Applying Filters")
    print("-"*80)
    
    filtered_customers = apply_filters(customers)
    
    if not filtered_customers:
        print("\n⚠️  No customers match your filter criteria")
        return
    
    if OUTPUT_OPTIONS["print_summary"]:
        print("\n" + "-"*80)
        print("STEP 3: Displaying Results")
        print("-"*80)
        print_summary(filtered_customers)
    
    print("\n" + "-"*80)
    print("STEP 4: Saving Results")
    print("-"*80)
    
    saved_files = []
    
    if OUTPUT_OPTIONS["save_json"]:
        json_file = save_to_json(filtered_customers)
        if json_file:
            saved_files.append(json_file)
    
    if OUTPUT_OPTIONS["save_csv"]:
        csv_file = save_to_csv(filtered_customers)
        if csv_file:
            saved_files.append(csv_file)
    
    print("\n" + "="*80)
    print("✅ PROCESS COMPLETED")
    print("="*80)
    print(f"\nTotal customers found: {len(filtered_customers)}")
    
    if saved_files:
        print(f"\nFiles created:")
        for file in saved_files:
            print(f"  • {file}")
    
    print("\n" + "="*80)

if __name__ == "__main__":
    main()
```
